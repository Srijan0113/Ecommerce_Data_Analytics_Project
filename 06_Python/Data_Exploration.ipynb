```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order ID"].nunique()
total_customers = df["Customer ID"].nunique()

print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
```

```python
# Category-wise Sales and Profit
category_analysis = df.groupby("Category")[["Sales", "Profit"]].sum()

category_analysis
```

```python
# Regioon-wise Sales ande Profit
region_analysis = df.groupby("Region")[["Sales", "Profit"]].sum()

region_analysis
```

```python
# Year-wise Sales and Profit
df["Order Date"] = pd.to_datetime(df["Order Date"])
year_analysis = df.groupby(df["Order Date"].dt.year)[["Sales", "Profit"]].sum()

year_analysis
```

```python
# Sub-category-wise Sales and Profit
subcategory_analysis = df.groupby("Sub-Category")[["Sales", "Profit"]].sum()

subcategory_analysis.sort_values("Sales", ascending=False)
```

```python
# Top 10 products by sales
top_products = (
    df.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products
```

# Top Products by Sales

The Canon imageCLASS 2200 Advanced Copier was the highest-selling product, generating approximately $61.6K in sales and approximately $25.2K in profit.

Some products contribute disproportionately to revenue and profit and can be important products for continued monitoring.

```python
# loss making products
loss_making_products = (
    df.groupby("Product Name")["Profit"]
    .sum()
    .sort_values(ascending=True)
    .head(10)
)
loss_making_products 

```

# Loss-Making Products

Several products generated negative total profit. The Cubify CubeX 3D Printer Double Head Print recorded the largest loss among the identified bottom-performing products, at approximately -$8.88K.

Loss-making products should be reviewed for pricing, discounting, cost, and demand-related factors before making decisions about their future.

```python
# Profit Margin
profit_margin = df["Profit"].sum() / df["Sales"].sum()

print("Profit Margin:", profit_margin)
print("Profit Margin (%):", profit_margin * 100)
```

```python
# Average Order Value (AOV)
aov = df["Sales"].sum() / df["Order ID"].nunique()

print("Average Order Value:", aov)
```

```python
# Summary
summary = {
    "Total Sales": df["Sales"].sum(),
    "Total Profit": df["Profit"].sum(),
    "Total Orders": df["Order ID"].nunique(),
    "Total Customers": df["Customer ID"].nunique(),
    "Average Order Value": df["Sales"].sum() / df["Order ID"].nunique(),
    "Profit Margin": df["Profit"].sum() / df["Sales"].sum()
}

summary
```
